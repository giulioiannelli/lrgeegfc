# Build correlation FC networks (single patient, dev mode)


## Parameters
Adjust these before running.

```python
PATIENT = 'Pat_02'
DEV_MODE = True

# Dev-mode defaults
DEV_BANDS = ['beta']
DEV_PHASES = ['rsPre']
DEV_FILTER_TIME = 20000  # samples; set None for full run

# Correlation settings
FILTER_TYPE = 'abs'
ZERO_DIAGONAL = True
FILTER_ORDER = 4
SAMPLE_RATE = 2048.0
CACHE_ROOT = 'data/corr_cache'
OVERWRITE = False
```

## Run computation
Uses cached matrices when available.

```python
from lrg_eegfc import compute_corr_for_patient

bands = DEV_BANDS if DEV_MODE else None
phases = DEV_PHASES if DEV_MODE else None
filter_time = DEV_FILTER_TIME if DEV_MODE else None

results = compute_corr_for_patient(
    PATIENT,
    bands=bands,
    phases=phases,
    filter_type=FILTER_TYPE,
    zero_diagonal=ZERO_DIAGONAL,
    filter_order=FILTER_ORDER,
    sample_rate=SAMPLE_RATE,
    filter_time=filter_time,
    cache_root=CACHE_ROOT,
    overwrite_cache=OVERWRITE,
    verbose=True,
)
```

## Summary table
Mean correlation per phase/band.

```python
import pandas as pd

rows = []
for band, phases_dict in results.items():
    for phase, result in phases_dict.items():
        if result is None:
            rows.append({'band': band, 'phase': phase, 'mean_corr': None, 'n_channels': None})
            continue
        rows.append({
            'band': band,
            'phase': phase,
            'mean_corr': result.mean_corr,
            'n_channels': result.n_channels,
        })

summary = pd.DataFrame(rows).sort_values(['band', 'phase'])
summary
```
