# 3D Brain Connectome Visualization

SEEG electrodes embedded in a 3D brain with functional connectivity edges.
Uses nilearn for proper brain-space visualization.

```python
%matplotlib inline
from lrgsglib.config.funcs import move_to_rootf
move_to_rootf(pathname="lrgeegfc")
from lrg_eegfc.notebook import *
```

## Parameters

```python
patient = "Pat_02"
phase = "rsPre"
band = "beta"
fc_method = "msc"
```

## Interactive 3D Brain Connectome (nilearn)

Electrodes in glass brain with all edges. Edge colors reflect connection weights.
- No thresholding by default
- Customize edge colormap with `edge_cmap`

```python
from lrg_eegfc.visuals import view_brain_connectome

# Full connectome - all edges, colored by weight
view = view_brain_connectome(
    patient,
    phase,
    band,
    fc_method=fc_method,
    edge_threshold=None,  # No thresholding - show all edges
    edge_cmap="coolwarm",  # Blue-red colormap
    linewidth=4.0,
    node_size=8.0,
    lrg_cache_root=Path("data/lrg_cache"),  # For cluster colors
)
view
```

## Different Edge Colormaps

Change `edge_cmap` to customize edge appearance:
- `"bwr"` (blue-white-red) - default, good for positive/negative
- `"coolwarm"` - similar but smoother
- `"viridis"` - perceptually uniform
- `"plasma"` - warm colors
- `"RdBu_r"` - diverging red-blue

```python
# Viridis colormap - good for all-positive weights
view_viridis = view_brain_connectome(
    patient,
    phase,
    band,
    fc_method=fc_method,
    edge_threshold=None,
    edge_cmap="viridis",
    symmetric_cmap=False,  # Not symmetric for viridis
    linewidth=3.0,
    node_size=6.0,
)
view_viridis
```

## With Edge Threshold (optional)

If you want to filter edges, use percentile or absolute threshold.

```python
# Top 10% strongest edges only
view_top = view_brain_connectome(
    patient,
    phase,
    band,
    fc_method=fc_method,
    edge_threshold="90%",  # Top 10% of edges
    edge_cmap="plasma",
    linewidth=6.0,
    node_size=10.0,
    title=f"{patient} {phase} {band} - Top 10% edges",
)
view_top
```
