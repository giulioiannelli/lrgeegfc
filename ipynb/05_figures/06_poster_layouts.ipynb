```python
import matplotlib.pyplot as plt

patient = list_patients(Path('data/stereoeeg_patients'))[0]
phase = PHASE_LABELS[0]
band = BRAIN_BANDS_NAMES[2]
fc_method = 'msc'

fig = plt.figure(figsize=(12, 8))
ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)
ax4 = fig.add_subplot(2, 2, 4)

corr = load_corr_matrix(patient, phase, band, filter_type='abs', zero_diagonal=True)
if corr is not None:
    ax1.imshow(corr, cmap='viridis', vmin=0, vmax=1)
    ax1.set_title('Correlation matrix')

msc = load_msc_matrix(patient, phase, band, sparsify='none', n_surrogates=0)
if msc is not None:
    ax2.imshow(msc, cmap='viridis', vmin=0, vmax=1)
    ax2.set_title('MSC matrix')

ax3.axis('off')
ax3.text(0.1, 0.5, 'Insert LRG panel here', fontsize=12)
ax4.axis('off')
ax4.text(0.1, 0.5, 'Insert reorganization heatmap here', fontsize=12)

plt.tight_layout()
plt.show()
```
