```python
import networkx as nx
from lrgsglib.core import get_giant_component
from lrg_eegfc.visuals.metastable import compute_clustering_across_tau, create_sankey_diagram

patient = list_patients(Path('data/stereoeeg_patients'))[0]
phase = PHASE_LABELS[1]
band = BRAIN_BANDS_NAMES[2]
fc_method = 'msc'

lrg = load_lrg_result(patient, phase, band, fc_method)
assert lrg is not None, 'Missing LRG cache'

fc_matrix = load_msc_matrix(patient, phase, band, sparsify='none', n_surrogates=0)
assert fc_matrix is not None, 'Missing MSC cache'

G = nx.from_numpy_array(fc_matrix)
Gcc = get_giant_component(G)

tau_values = lrg.entropy_tau[::5]
partitions, _ = compute_clustering_across_tau(lrg.linkage_matrix, tau_values, Gcc)

fig = create_sankey_diagram(partitions, tau_values)
fig
```
