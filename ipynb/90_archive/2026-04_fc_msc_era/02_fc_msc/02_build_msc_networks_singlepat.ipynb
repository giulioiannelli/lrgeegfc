# Build MSC FC networks (single patient, dev mode)


## Parameters
Adjust these before running.

```python
PATIENT = 'Pat_02'
DEV_MODE = True

# Dev-mode defaults
DEV_BANDS = ['beta']
DEV_PHASES = ['rsPre']
DEV_FILTER_TIME = 20000  # samples; set None for full run

# MSC settings
SPARSIFY = 'none'  # 'none' (dev) or 'soft' (validated)
N_SURROGATES = 0    # raise for final validated runs
NPERSEG = 1024
BATCH_SIZE = 64
SAMPLE_RATE = 2048.0
CACHE_ROOT = 'data/msc_cache'
OVERWRITE = False
```

## Run computation
Uses cached matrices when available.

```python
from lrg_eegfc import compute_msc_for_patient

bands = DEV_BANDS if DEV_MODE else None
phases = DEV_PHASES if DEV_MODE else None
filter_time = DEV_FILTER_TIME if DEV_MODE else None

results = compute_msc_for_patient(
    PATIENT,
    bands=bands,
    phases=phases,
    sparsify=SPARSIFY,
    n_surrogates=N_SURROGATES,
    nperseg=NPERSEG,
    batch_size=BATCH_SIZE,
    sample_rate=SAMPLE_RATE,
    filter_time=filter_time,
    cache_root=CACHE_ROOT,
    overwrite_cache=OVERWRITE,
    verbose=True,
)
```

## Summary table
Mean MSC per phase/band.

```python
import pandas as pd

rows = []
for band, phases_dict in results.items():
    for phase, result in phases_dict.items():
        if result is None:
            rows.append({'band': band, 'phase': phase, 'mean_msc': None, 'n_channels': None})
            continue
        rows.append({
            'band': band,
            'phase': phase,
            'mean_msc': result.mean_msc,
            'n_channels': result.n_channels,
        })

summary = pd.DataFrame(rows).sort_values(['band', 'phase'])
summary
```
