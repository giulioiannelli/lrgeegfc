```python
import networkx as nx
from lrgsglib.core import entropy

# random graph sandbox
n = 50
G = nx.erdos_renyi_graph(n, 0.1, seed=42)
for u, v in G.edges():
    G[u][v]['weight'] = 0.5

adj = nx.to_numpy_array(G)

# threshold sweep
thresholds = np.linspace(0.1, 0.9, 5)
for th in thresholds:
    adj_th = adj.copy()
    adj_th[adj_th < th] = 0
    Gth = nx.from_numpy_array(adj_th)
    Sm1, Csp, VarL, tt = entropy(Gth)
    print('th', th, 'entropy len', len(Sm1))
```
