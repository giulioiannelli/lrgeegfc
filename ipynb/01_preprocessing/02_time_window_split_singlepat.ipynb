# Time-window split (single patient)

Define time windows with overlap and inspect coverage for one patient/phase.

```python
%matplotlib inline
from lrgsglib.config.funcs import move_to_rootf
move_to_rootf(pathname="lrg_eegfc")
from lrg_eegfc.notebook import *
```

```python
DATA_ROOT = Path('data/stereoeeg_patients')
patients = list_patients(DATA_ROOT)
assert patients, 'No patients found under data/stereoeeg_patients'

patient = patients[0]
phase = PHASE_LABELS[0]

recording = load_patient_dataset_robust(patient, DATA_ROOT, phases=[phase])[phase]
timeseries = recording.timeseries
fs = float(recording.parameters.get('fs'))

n_channels, n_samples = timeseries.shape
print('channels', n_channels, 'samples', n_samples, 'fs', fs)

# window defaults
window_sec = 10.0
overlap = 0.25
window_len = int(window_sec * fs)
step = int(window_len * (1.0 - overlap))

n_windows = max(1, 1 + (n_samples - window_len) // step)
print('window_len', window_len, 'step', step, 'n_windows', n_windows)

# quick sanity: T >> N for MSC stability
print('T / N', n_samples / n_channels)
```

```python
import matplotlib.pyplot as plt

channel_idx = 0
fig, ax = plt.subplots(figsize=(10, 3))
ax.plot(timeseries[channel_idx], linewidth=0.5)
for i in range(min(n_windows, 5)):
    start = i * step
    ax.axvline(start, color='red', alpha=0.3)
ax.set_title(f'{patient} {phase} channel {channel_idx}')
ax.set_xlabel('sample')
plt.tight_layout()
plt.show()
```
